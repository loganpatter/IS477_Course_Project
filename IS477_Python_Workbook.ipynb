```python
# Uses my API Key and the series ID along with the requests library to obtain the data from the FRED API, turning it into a json file
API_KEY = "6854f98f39927bf9f410ddc3a8fd4c85" ## Won't hardcode in final project files
series_id = "LAUCN481130000000003A" ## Also won't hardcode in final project files

url = "https://api.stlouisfed.org/fred/series/observations"
params = {
    "series_id": series_id,
    "api_key": API_KEY,
    "file_type": "json",
    "observation_start": "2020-01-01", #Change depending on start date needed
    "observation_end": "2025-01-01"
}

resp = requests.get(url, params=params)

data = resp.json()
```

```python
# Uses the json stored data to create a pandas dataframe of the API data
obs = data["observations"]
dallas_unemploy_df = pd.DataFrame(obs)

# Convert types for better analysis and more continuity between datasets once merged
dallas_unemploy_df["date"] = pd.to_datetime(dallas_unemploy_df["date"])
dallas_unemploy_df["value"] = pd.to_numeric(dallas_unemploy_df["value"], errors="coerce")

dallas_unemploy_df.head()
```

```python
dallas_unemploy_df['County'] = "Dallas" #Adding identifiers to the County data 
```
